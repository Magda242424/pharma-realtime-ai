```python
# ============================================================
# PHARMA — CREATE IoT DATA GENERATOR
# ============================================================

generate_data_content = '''
from __future__ import annotations

from pathlib import Path

import numpy as np
import pandas as pd


ROOT = Path(__file__).resolve().parents[1]

RAW_DIR = ROOT / "data" / "raw"


def generate_sensor_data(
    n_rows: int = 5000,
    n_equipment: int = 3,
    seed: int = 42,
) -> pd.DataFrame:

    rng = np.random.default_rng(seed)

    timestamps = pd.date_range(
        start="2025-01-01",
        periods=n_rows,
        freq="min",
    )

    equipment_ids = [
        f"EQ-{i:03d}"
        for i in range(1, n_equipment + 1)
    ]

    equipment = rng.choice(
        equipment_ids,
        size=n_rows,
    )

    time_index = np.arange(n_rows)

    temperature = (
        25
        + 1.5 * np.sin(time_index / 120)
        + rng.normal(0, 0.3, n_rows)
    )

    ph = (
        7.0
        + 0.15 * np.sin(time_index / 90)
        + rng.normal(0, 0.03, n_rows)
    )

    dissolved_oxygen = (
        7.5
        - 0.015 * (temperature - 25)
        + 0.3 * np.sin(time_index / 80)
        + rng.normal(0, 0.12, n_rows)
    )

    agitation_speed = (
        150
        + 10 * np.sin(time_index / 100)
        + rng.normal(0, 3, n_rows)
    )

    pressure = (
        1.2
        + 0.05 * np.sin(time_index / 150)
        + rng.normal(0, 0.01, n_rows)
    )

    flow_rate = (
        100
        + 5 * np.sin(time_index / 110)
        + rng.normal(0, 1.5, n_rows)
    )

    df = pd.DataFrame(
        {
            "timestamp": timestamps,
            "batch_id": "BATCH-001",
            "equipment_id": equipment,
            "temperature": temperature,
            "ph": ph,
            "dissolved_oxygen": dissolved_oxygen,
            "agitation_speed": agitation_speed,
            "pressure": pressure,
            "flow_rate": flow_rate,
            "source_system": "iot_simulator",
        }
    )

    # --------------------------------------------------------
    # Synthetic anomalies
    # --------------------------------------------------------

    anomaly_indices = rng.choice(
        n_rows,
        size=max(1, n_rows // 100),
        replace=False,
    )

    df.loc[
        anomaly_indices,
        "dissolved_oxygen",
    ] += rng.normal(
        2.0,
        0.5,
        len(anomaly_indices),
    )

    return df


def save_raw_data(df: pd.DataFrame) -> Path:

    RAW_DIR.mkdir(
        parents=True,
        exist_ok=True,
    )

    output = RAW_DIR / "sensor_data.csv"

    df.to_csv(
        output,
        index=False,
    )

    return output


if __name__ == "__main__":

    df = generate_sensor_data()

    path = save_raw_data(df)

    print(f"Generated {len(df)} rows")
    print(f"Saved to: {path}")
'''.strip() + "\n"

GENERATE_DATA_FILE = (
    SCRIPTS_DIR / "generate_data.py"
)

GENERATE_DATA_FILE.write_text(
    generate_data_content,
    encoding="utf-8",
)

print("=" * 60)
print("PHARMA — IoT DATA GENERATOR")
print("=" * 60)
print()
print("Created:", GENERATE_DATA_FILE)
print()
print("Generator file: OK")
```

```python
# ============================================================
# PHARMA — CREATE DATA QUALITY SCRIPT
# ============================================================

validate_data_content = '''
from __future__ import annotations

from pathlib import Path

import pandas as pd


ROOT = Path(__file__).resolve().parents[1]

RAW_FILE = (
    ROOT
    / "data"
    / "raw"
    / "sensor_data.csv"
)

PROCESSED_DIR = (
    ROOT
    / "data"
    / "processed"
)


REQUIRED_COLUMNS = [
    "timestamp",
    "batch_id",
    "equipment_id",
    "temperature",
    "ph",
    "dissolved_oxygen",
    "agitation_speed",
    "pressure",
    "flow_rate",
    "source_system",
]


def validate_schema(df: pd.DataFrame) -> None:

    missing = [
        column
        for column in REQUIRED_COLUMNS
        if column not in df.columns
    ]

    if missing:
        raise ValueError(
            f"Missing required columns: {missing}"
        )


def clean_data(df: pd.DataFrame) -> pd.DataFrame:

    validate_schema(df)

    result = df.copy()

    result["timestamp"] = pd.to_datetime(
        result["timestamp"],
        errors="coerce",
    )

    numeric_columns = [
        "temperature",
        "ph",
        "dissolved_oxygen",
        "agitation_speed",
        "pressure",
        "flow_rate",
    ]

    for column in numeric_columns:

        result[column] = pd.to_numeric(
            result[column],
            errors="coerce",
        )

    result = result.dropna(
        subset=[
            "timestamp",
            "equipment_id",
            "dissolved_oxygen",
        ]
    )

    result = result.drop_duplicates(
        subset=[
            "timestamp",
            "equipment_id",
        ]
    )

    result = result.sort_values(
        [
            "equipment_id",
            "timestamp",
        ]
    )

    return result.reset_index(drop=True)


def save_processed_data(
    df: pd.DataFrame,
) -> Path:

    PROCESSED_DIR.mkdir(
        parents=True,
        exist_ok=True,
    )

    output = (
        PROCESSED_DIR
        / "sensor_data_clean.csv"
    )

    df.to_csv(
        output,
        index=False,
    )

    return output


if __name__ == "__main__":

    if not RAW_FILE.exists():
        raise FileNotFoundError(
            f"Raw data not found: {RAW_FILE}"
        )

    df = pd.read_csv(RAW_FILE)

    clean = clean_data(df)

    output = save_processed_data(clean)

    print(f"Clean dataset: {len(clean)} rows")
    print(f"Saved to: {output}")
'''.strip() + "\n"

VALIDATE_DATA_FILE = (
    SCRIPTS_DIR / "validate_data.py"
)

VALIDATE_DATA_FILE.write_text(
    validate_data_content,
    encoding="utf-8",
)

print("=" * 60)
print("PHARMA — DATA QUALITY SCRIPT")
print("=" * 60)
print()
print("Created:", VALIDATE_DATA_FILE)
print()
print("Data quality script: OK")
```

```python
# ============================================================
# PHARMA — MODEL TRAINING SCRIPT
# ============================================================

from pathlib import Path

TRAIN_FILE = PROJECT_DIR / "src" / "models" / "train.py"

TRAIN_CODE = r'''
from __future__ import annotations

import os
from pathlib import Path

import joblib
import mlflow
import pandas as pd

from sklearn.ensemble import IsolationForest, RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

from src.features.feature_definitions import (
    ANOMALY_FEATURES,
    FORECASTING_FEATURES,
    TARGET_COLUMN,
)


# ============================================================
# PATHS
# ============================================================

ROOT = Path(__file__).resolve().parents[2]

DATA_FILE = ROOT / "data" / "processed" / "features.csv"
MODEL_DIR = ROOT / "models"

MODEL_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# ANOMALY DETECTION
# ============================================================

def train_anomaly_model(df: pd.DataFrame):

    training_df = df[ANOMALY_FEATURES].dropna()

    model = IsolationForest(
        n_estimators=200,
        contamination=0.02,
        random_state=42,
    )

    model.fit(training_df)

    output = MODEL_DIR / "anomaly_model.joblib"

    joblib.dump(model, output)

    return model, output


# ============================================================
# FORECASTING
# ============================================================

def train_forecasting_model(df: pd.DataFrame):

    training_df = df[
        FORECASTING_FEATURES + [TARGET_COLUMN]
    ].dropna()

    X = training_df[FORECASTING_FEATURES]
    y = training_df[TARGET_COLUMN]

    # Chronological split
    split = int(len(X) * 0.8)

    X_train = X.iloc[:split]
    X_test = X.iloc[split:]

    y_train = y.iloc[:split]
    y_test = y.iloc[split:]

    model = RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1,
    )

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    mae = mean_absolute_error(
        y_test,
        predictions,
    )

    rmse = mean_squared_error(
        y_test,
        predictions,
    ) ** 0.5

    output = MODEL_DIR / "forecasting_model.joblib"

    joblib.dump(model, output)

    return model, output, mae, rmse


# ============================================================
# MAIN TRAINING PIPELINE
# ============================================================

def main():

    print("=" * 60)
    print("PHARMA — MODEL TRAINING")
    print("=" * 60)

    print()
    print("Input file:")
    print(DATA_FILE)

    if not DATA_FILE.exists():
        raise FileNotFoundError(
            f"Feature dataset not found: {DATA_FILE}"
        )

    df = pd.read_csv(DATA_FILE)

    print()
    print(f"Input rows: {len(df)}")

    # --------------------------------------------------------
    # MLflow configuration
    # --------------------------------------------------------

    tracking_uri = os.getenv(
        "MLFLOW_TRACKING_URI",
        "http://localhost:5000",
    )

    mlflow.set_tracking_uri(tracking_uri)

    mlflow.set_experiment(
        "pharma-realtime-ai"
    )

    # --------------------------------------------------------
    # MLflow run
    # --------------------------------------------------------

    with mlflow.start_run(
        run_name="pharma-model-training"
    ):

        # ----------------------------------------------------
        # Anomaly model
        # ----------------------------------------------------

        anomaly_model, anomaly_path = train_anomaly_model(df)

        mlflow.log_param(
            "anomaly_model",
            "IsolationForest",
        )

        mlflow.log_param(
            "anomaly_features",
            len(ANOMALY_FEATURES),
        )

        mlflow.log_param(
            "anomaly_estimators",
            200,
        )

        # ----------------------------------------------------
        # Forecasting model
        # ----------------------------------------------------

        (
            forecasting_model,
            forecasting_path,
            mae,
            rmse,
        ) = train_forecasting_model(df)

        mlflow.log_param(
            "forecasting_model",
            "RandomForestRegressor",
        )

        mlflow.log_param(
            "forecasting_features",
            len(FORECASTING_FEATURES),
        )

        mlflow.log_param(
            "forecasting_estimators",
            200,
        )

        mlflow.log_metric(
            "mae",
            mae,
        )

        mlflow.log_metric(
            "rmse",
            rmse,
        )

        print()
        print("Anomaly model:")
        print(f"  Saved to: {anomaly_path}")

        print()
        print("Forecasting model:")
        print(f"  Saved to: {forecasting_path}")

        print()
        print("Forecast metrics:")
        print(f"  MAE  : {mae:.4f}")
        print(f"  RMSE : {rmse:.4f}")

    print()
    print("=" * 60)
    print("MODEL TRAINING: OK")
    print("=" * 60)


if __name__ == "__main__":
    main()
'''

TRAIN_FILE.parent.mkdir(parents=True, exist_ok=True)
TRAIN_FILE.write_text(TRAIN_CODE, encoding="utf-8")

print("=" * 60)
print("PHARMA — MODEL TRAINING SCRIPT")
print("=" * 60)
print()
print(f"Created: {TRAIN_FILE}")
print()
print("Training script: OK")
```

```python
# ============================================================
# PHARMA — MODEL INFERENCE
# ============================================================

from pathlib import Path
import sys

import joblib
import pandas as pd

# ------------------------------------------------------------
# PROJECT
# ------------------------------------------------------------

PROJECT_ROOT = Path(r"C:\Users\zosia\pharma")

DATA_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "features.csv"
)

MODEL_DIR = PROJECT_ROOT / "models"

ANOMALY_MODEL_FILE = (
    MODEL_DIR
    / "anomaly_model.joblib"
)

FORECASTING_MODEL_FILE = (
    MODEL_DIR
    / "forecasting_model.joblib"
)


# ------------------------------------------------------------
# HEADER
# ------------------------------------------------------------

print("=" * 60)
print("PHARMA — MODEL INFERENCE")
print("=" * 60)


# ------------------------------------------------------------
# PROJECT CHECK
# ------------------------------------------------------------

print()
print("Project:")
print(PROJECT_ROOT)

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(
        f"Project directory not found: {PROJECT_ROOT}"
    )


# ------------------------------------------------------------
# DATA CHECK
# ------------------------------------------------------------

print()
print("Input data:")
print(DATA_FILE)

if not DATA_FILE.exists():
    raise FileNotFoundError(
        f"Feature file not found: {DATA_FILE}"
    )


# ------------------------------------------------------------
# MODEL CHECK
# ------------------------------------------------------------

print()
print("Anomaly model:")
print(ANOMALY_MODEL_FILE)

if not ANOMALY_MODEL_FILE.exists():
    raise FileNotFoundError(
        f"Anomaly model not found: {ANOMALY_MODEL_FILE}"
    )

print()
print("Forecasting model:")
print(FORECASTING_MODEL_FILE)

if not FORECASTING_MODEL_FILE.exists():
    raise FileNotFoundError(
        f"Forecasting model not found: {FORECASTING_MODEL_FILE}"
    )


# ------------------------------------------------------------
# LOAD FEATURE CONTRACT
# ------------------------------------------------------------

print()
print("-" * 60)
print("FEATURE CONTRACT")
print("-" * 60)

sys.path.insert(
    0,
    str(PROJECT_ROOT),
)

from src.features.feature_definitions import (
    ANOMALY_FEATURES,
    FORECASTING_FEATURES,
    TARGET_COLUMN,
)

print()
print("Anomaly features:")
print(ANOMALY_FEATURES)

print()
print("Forecasting features:")
print(FORECASTING_FEATURES)

print()
print("Target:")
print(TARGET_COLUMN)


# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

print()
print("-" * 60)
print("LOADING DATA")
print("-" * 60)

df = pd.read_csv(DATA_FILE)

print()
print("Input rows:", len(df))
print("Input columns:", len(df.columns))


# ------------------------------------------------------------
# CHECK REQUIRED FEATURES
# ------------------------------------------------------------

required_features = list(
    set(
        ANOMALY_FEATURES
        + FORECASTING_FEATURES
    )
)

missing_features = [
    feature
    for feature in required_features
    if feature not in df.columns
]

if missing_features:
    raise ValueError(
        "Missing inference features: "
        f"{missing_features}"
    )

print()
print("Required inference features: OK")


# ------------------------------------------------------------
# LOAD MODELS
# ------------------------------------------------------------

print()
print("-" * 60)
print("LOADING MODELS")
print("-" * 60)

anomaly_model = joblib.load(
    ANOMALY_MODEL_FILE
)

forecasting_model = joblib.load(
    FORECASTING_MODEL_FILE
)

print()
print("Anomaly model loaded: OK")
print("Forecasting model loaded: OK")


# ------------------------------------------------------------
# PREPARE ANOMALY DATA
# ------------------------------------------------------------

anomaly_data = df[
    ANOMALY_FEATURES
].dropna()

print()
print("Anomaly inference rows:")
print(len(anomaly_data))


# ------------------------------------------------------------
# ANOMALY INFERENCE
# ------------------------------------------------------------

print()
print("-" * 60)
print("ANOMALY DETECTION")
print("-" * 60)

anomaly_predictions = (
    anomaly_model.predict(
        anomaly_data
    )
)

anomaly_scores = (
    anomaly_model.decision_function(
        anomaly_data
    )
)

anomaly_results = anomaly_data.copy()

anomaly_results[
    "anomaly_prediction"
] = anomaly_predictions

anomaly_results[
    "anomaly_score"
] = anomaly_scores

anomaly_results[
    "is_anomaly"
] = (
    anomaly_predictions == -1
)

anomaly_count = int(
    anomaly_results[
        "is_anomaly"
    ].sum()
)

print()
print("Rows analysed:", len(anomaly_results))
print("Anomalies detected:", anomaly_count)


# ------------------------------------------------------------
# PREPARE FORECASTING DATA
# ------------------------------------------------------------

forecasting_data = df[
    FORECASTING_FEATURES
].dropna()

print()
print("Forecasting inference rows:")
print(len(forecasting_data))


# ------------------------------------------------------------
# FORECASTING INFERENCE
# ------------------------------------------------------------

print()
print("-" * 60)
print("DISSOLVED OXYGEN FORECASTING")
print("-" * 60)

forecast_predictions = (
    forecasting_model.predict(
        forecasting_data
    )
)

forecast_results = forecasting_data.copy()

forecast_results[
    "predicted_dissolved_oxygen"
] = forecast_predictions


# ------------------------------------------------------------
# DISPLAY FORECAST RESULTS
# ------------------------------------------------------------

print()
print("Forecast predictions generated:")
print(
    len(
        forecast_results
    )
)

print()
print("First 5 predictions:")

print(
    forecast_results[
        ["predicted_dissolved_oxygen"]
    ].head()
)


# ------------------------------------------------------------
# COMBINED INFERENCE OUTPUT
# ------------------------------------------------------------

print()
print("-" * 60)
print("INFERENCE SUMMARY")
print("-" * 60)

print()
print(
    "Total input rows:",
    len(df),
)

print(
    "Anomaly rows:",
    len(anomaly_results),
)

print(
    "Detected anomalies:",
    anomaly_count,
)

print(
    "Forecast rows:",
    len(forecast_results),
)

print()
print("Forecast prediction statistics:")

print(
    forecast_results[
        "predicted_dissolved_oxygen"
    ].describe()
)


# ------------------------------------------------------------
# SAMPLE ANOMALIES
# ------------------------------------------------------------

print()
print("-" * 60)
print("ANOMALY SAMPLE")
print("-" * 60)

if anomaly_count > 0:

    print(
        anomaly_results[
            anomaly_results[
                "is_anomaly"
            ]
        ].head(10)
    )

else:

    print(
        "No anomalies detected."
    )


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print()
print("=" * 60)
print("PHARMA — MODEL INFERENCE: SUCCESS")
print("=" * 60)
```

```python
# ============================================================
# PHARMA — FASTAPI
# ============================================================

from pathlib import Path
import subprocess
import sys
import textwrap
import time
import requests

# ============================================================
# PROJECT
# ============================================================

ROOT = Path(r"C:\Users\zosia\pharma")

API_DIR = ROOT / "src" / "api"
API_FILE = API_DIR / "main.py"

API_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 60)
print("PHARMA — FASTAPI SETUP")
print("=" * 60)

print()
print("Project:")
print(ROOT)

print()
print("API file:")
print(API_FILE)

# ============================================================
# CREATE API
# ============================================================

api_code = r'''
from __future__ import annotations

from pathlib import Path

import joblib
import pandas as pd
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel


# ============================================================
# PATHS
# ============================================================

ROOT = Path(__file__).resolve().parents[2]

MODEL_DIR = ROOT / "models"

ANOMALY_MODEL_FILE = MODEL_DIR / "anomaly_model.joblib"
FORECASTING_MODEL_FILE = MODEL_DIR / "forecasting_model.joblib"


# ============================================================
# FEATURE CONTRACT
# ============================================================

ANOMALY_FEATURES = [
    "temperature",
    "ph",
    "dissolved_oxygen",
    "agitation_speed",
    "pressure",
    "flow_rate",
    "dissolved_oxygen_lag_1",
    "dissolved_oxygen_lag_2",
    "dissolved_oxygen_lag_3",
    "dissolved_oxygen_lag_5",
    "dissolved_oxygen_lag_10",
    "dissolved_oxygen_lag_15",
    "dissolved_oxygen_lag_30",
]

FORECASTING_FEATURES = [
    "dissolved_oxygen_lag_1",
    "dissolved_oxygen_lag_2",
    "dissolved_oxygen_lag_3",
    "dissolved_oxygen_lag_5",
    "dissolved_oxygen_lag_10",
    "dissolved_oxygen_lag_15",
    "dissolved_oxygen_lag_30",
]

TARGET_COLUMN = "dissolved_oxygen"


# ============================================================
# LOAD MODELS
# ============================================================

if not ANOMALY_MODEL_FILE.exists():
    raise FileNotFoundError(
        f"Anomaly model not found: {ANOMALY_MODEL_FILE}"
    )

if not FORECASTING_MODEL_FILE.exists():
    raise FileNotFoundError(
        f"Forecasting model not found: {FORECASTING_MODEL_FILE}"
    )


anomaly_model = joblib.load(ANOMALY_MODEL_FILE)
forecasting_model = joblib.load(FORECASTING_MODEL_FILE)


# ============================================================
# FASTAPI APPLICATION
# ============================================================

app = FastAPI(
    title="PHARMA Real-Time AI API",
    version="1.0.0",
    description="API for pharmaceutical sensor anomaly detection and dissolved oxygen forecasting.",
)


# ============================================================
# REQUEST MODEL
# ============================================================

class SensorReading(BaseModel):
    temperature: float
    ph: float
    dissolved_oxygen: float
    agitation_speed: float
    pressure: float
    flow_rate: float

    dissolved_oxygen_lag_1: float
    dissolved_oxygen_lag_2: float
    dissolved_oxygen_lag_3: float
    dissolved_oxygen_lag_5: float
    dissolved_oxygen_lag_10: float
    dissolved_oxygen_lag_15: float
    dissolved_oxygen_lag_30: float


# ============================================================
# HEALTH CHECK
# ============================================================

@app.get("/")
def root():
    return {
        "service": "PHARMA Real-Time AI API",
        "status": "ok",
        "version": "1.0.0",
    }


@app.get("/health")
def health():
    return {
        "status": "healthy",
        "anomaly_model": ANOMALY_MODEL_FILE.exists(),
        "forecasting_model": FORECASTING_MODEL_FILE.exists(),
    }


# ============================================================
# FEATURE VALIDATION
# ============================================================

def build_dataframe(reading: SensorReading) -> pd.DataFrame:
    data = reading.model_dump()

    missing = [
        feature
        for feature in ANOMALY_FEATURES
        if feature not in data
    ]

    if missing:
        raise HTTPException(
            status_code=422,
            detail={
                "message": "Missing required features",
                "features": missing,
            },
        )

    return pd.DataFrame([data])


# ============================================================
# ANOMALY DETECTION
# ============================================================

@app.post("/predict/anomaly")
def predict_anomaly(reading: SensorReading):

    df = build_dataframe(reading)

    X = df[ANOMALY_FEATURES]

    prediction = int(anomaly_model.predict(X)[0])
    score = float(anomaly_model.decision_function(X)[0])

    return {
        "anomaly_prediction": prediction,
        "is_anomaly": prediction == -1,
        "anomaly_score": score,
    }


# ============================================================
# DISSOLVED OXYGEN FORECAST
# ============================================================

@app.post("/predict/forecast")
def predict_forecast(reading: SensorReading):

    df = build_dataframe(reading)

    X = df[FORECASTING_FEATURES]

    prediction = float(forecasting_model.predict(X)[0])

    return {
        "predicted_dissolved_oxygen": prediction,
    }


# ============================================================
# COMBINED PREDICTION
# ============================================================

@app.post("/predict")
def predict(reading: SensorReading):

    df = build_dataframe(reading)

    anomaly_X = df[ANOMALY_FEATURES]
    forecast_X = df[FORECASTING_FEATURES]

    anomaly_prediction = int(
        anomaly_model.predict(anomaly_X)[0]
    )

    anomaly_score = float(
        anomaly_model.decision_function(anomaly_X)[0]
    )

    forecast_prediction = float(
        forecasting_model.predict(forecast_X)[0]
    )

    return {
        "anomaly_prediction": anomaly_prediction,
        "is_anomaly": anomaly_prediction == -1,
        "anomaly_score": anomaly_score,
        "predicted_dissolved_oxygen": forecast_prediction,
    }
'''

API_FILE.write_text(
    textwrap.dedent(api_code),
    encoding="utf-8"
)

print()
print("Created:")
print(API_FILE)

# ============================================================
# VALIDATE PYTHON SYNTAX
# ============================================================

print()
print("-" * 60)
print("API SYNTAX VALIDATION")
print("-" * 60)

syntax_result = subprocess.run(
    [
        sys.executable,
        "-m",
        "py_compile",
        str(API_FILE),
    ],
    cwd=str(ROOT),
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
)

print("Return code:", syntax_result.returncode)

if syntax_result.stdout:
    print(syntax_result.stdout)

if syntax_result.stderr:
    print(syntax_result.stderr)

if syntax_result.returncode != 0:
    raise RuntimeError(
        "FastAPI syntax validation failed."
    )

print()
print("API syntax: OK")

# ============================================================
# START FASTAPI
# ============================================================

print()
print("=" * 60)
print("STARTING PHARMA FASTAPI")
print("=" * 60)

server_command = [
    sys.executable,
    "-m",
    "uvicorn",
    "src.api.main:app",
    "--host",
    "127.0.0.1",
    "--port",
    "8000",
]

server = subprocess.Popen(
    server_command,
    cwd=str(ROOT),
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    encoding="utf-8",
    errors="replace",
)

# ============================================================
# WAIT FOR SERVER
# ============================================================

print()
print("Waiting for FastAPI...")

server_ready = False

for _ in range(20):
    time.sleep(1)

    try:
        response = requests.get(
            "http://127.0.0.1:8000/health",
            timeout=2,
        )

        if response.status_code == 200:
            server_ready = True
            break

    except requests.RequestException:
        pass


if not server_ready:
    print()
    print("FASTAPI STARTUP FAILED")

    if server.poll() is not None:
        output, _ = server.communicate(timeout=2)
        print(output)

    raise RuntimeError(
        "FastAPI server did not become ready."
    )


# ============================================================
# HEALTH TEST
# ============================================================

health_response = requests.get(
    "http://127.0.0.1:8000/health",
    timeout=5,
)

print()
print("-" * 60)
print("HEALTH CHECK")
print("-" * 60)

print("Status code:", health_response.status_code)
print("Response:", health_response.json())

if health_response.status_code != 200:
    raise RuntimeError(
        "FastAPI health check failed."
    )

# ============================================================
# FINAL RESULT
# ============================================================

print()
print("=" * 60)
print("PHARMA — FASTAPI: SUCCESS")
print("=" * 60)

print()
print("API:")
print("http://127.0.0.1:8000")

print()
print("Swagger:")
print("http://127.0.0.1:8000/docs")

print()
print("Health:")
print("http://127.0.0.1:8000/health")

print()
print("Endpoints:")
print("POST /predict")
print("POST /predict/anomaly")
print("POST /predict/forecast")

print()
print("FastAPI server is running.")
```

```python
# ============================================================
# PHARMA — STREAMLIT DASHBOARD SETUP
# ============================================================

from pathlib import Path
import subprocess
import sys
import time
import requests

# ============================================================
# PROJECT
# ============================================================

ROOT = Path(r"C:\Users\zosia\pharma")

DASHBOARD_FILE = (
    ROOT
    / "src"
    / "dashboard"
    / "app.py"
)

API_URL = "http://127.0.0.1:8000"

print("=" * 60)
print("PHARMA — STREAMLIT DASHBOARD SETUP")
print("=" * 60)

print()
print("Project:")
print(ROOT)

# ============================================================
# CHECK PROJECT
# ============================================================

if not ROOT.exists():
    raise FileNotFoundError(
        f"Project directory not found: {ROOT}"
    )

print()
print("Project exists: OK")

# ============================================================
# CHECK STREAMLIT
# ============================================================

print()
print("-" * 60)
print("STREAMLIT CHECK")
print("-" * 60)

streamlit_check = subprocess.run(
    [
        sys.executable,
        "-m",
        "streamlit",
        "--version",
    ],
    cwd=ROOT,
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
)

print("Return code:", streamlit_check.returncode)

if streamlit_check.returncode != 0:
    print()
    print("ERROR:")
    print(streamlit_check.stderr)

    raise RuntimeError(
        "Streamlit is not installed in the current Python environment."
    )

print(
    "Version:",
    streamlit_check.stdout.strip(),
)

print()
print("Streamlit: OK")

# ============================================================
# CREATE DASHBOARD DIRECTORY
# ============================================================

DASHBOARD_FILE.parent.mkdir(
    parents=True,
    exist_ok=True,
)

print()
print("Dashboard directory: OK")

# ============================================================
# DASHBOARD APPLICATION
# ============================================================

dashboard_code = r'''
from pathlib import Path

import pandas as pd
import requests
import streamlit as st


# ============================================================
# CONFIGURATION
# ============================================================

ROOT = Path(__file__).resolve().parents[2]

FEATURES_FILE = (
    ROOT
    / "data"
    / "processed"
    / "features.csv"
)

API_URL = "http://127.0.0.1:8000"


# ============================================================
# PAGE CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="PHARMA — Real-Time AI",
    page_icon="⚗️",
    layout="wide",
)


# ============================================================
# TITLE
# ============================================================

st.title("PHARMA — Real-Time AI")
st.caption(
    "Pharmaceutical process monitoring — "
    "anomaly detection and dissolved oxygen forecasting"
)


# ============================================================
# LOAD DATA
# ============================================================

@st.cache_data
def load_data():

    if not FEATURES_FILE.exists():
        raise FileNotFoundError(
            f"Features file not found: {FEATURES_FILE}"
        )

    return pd.read_csv(FEATURES_FILE)


# ============================================================
# API HEALTH
# ============================================================

def check_api():

    try:

        response = requests.get(
            f"{API_URL}/health",
            timeout=3,
        )

        if response.status_code == 200:
            return response.json()

        return None

    except requests.RequestException:
        return None


# ============================================================
# API PREDICTION
# ============================================================

def predict(row):

    payload = {
        "temperature": float(row["temperature"]),
        "ph": float(row["ph"]),
        "dissolved_oxygen": float(row["dissolved_oxygen"]),
        "agitation_speed": float(row["agitation_speed"]),
        "pressure": float(row["pressure"]),
        "flow_rate": float(row["flow_rate"]),
        "dissolved_oxygen_lag_1": float(
            row["dissolved_oxygen_lag_1"]
        ),
        "dissolved_oxygen_lag_2": float(
            row["dissolved_oxygen_lag_2"]
        ),
        "dissolved_oxygen_lag_3": float(
            row["dissolved_oxygen_lag_3"]
        ),
        "dissolved_oxygen_lag_5": float(
            row["dissolved_oxygen_lag_5"]
        ),
        "dissolved_oxygen_lag_10": float(
            row["dissolved_oxygen_lag_10"]
        ),
        "dissolved_oxygen_lag_15": float(
            row["dissolved_oxygen_lag_15"]
        ),
        "dissolved_oxygen_lag_30": float(
            row["dissolved_oxygen_lag_30"]
        ),
    }

    response = requests.post(
        f"{API_URL}/predict",
        json=payload,
        timeout=10,
    )

    response.raise_for_status()

    return response.json()


# ============================================================
# LOAD DATA
# ============================================================

try:

    df = load_data()

except Exception as exc:

    st.error(
        f"Unable to load feature data: {exc}"
    )

    st.stop()


# ============================================================
# SIDEBAR
# ============================================================

st.sidebar.header("PHARMA Controls")

equipment_options = sorted(
    df["equipment_id"].dropna().unique()
)

selected_equipment = st.sidebar.selectbox(
    "Equipment",
    ["All"] + equipment_options,
)

# ============================================================
# FILTER DATA
# ============================================================

if selected_equipment == "All":

    filtered_df = df.copy()

else:

    filtered_df = df[
        df["equipment_id"] == selected_equipment
    ].copy()


# ============================================================
# API STATUS
# ============================================================

api_status = check_api()

st.sidebar.subheader("System Status")

if api_status:

    st.sidebar.success("FastAPI: ONLINE")

    if api_status.get("anomaly_model"):

        st.sidebar.success(
            "Anomaly model: LOADED"
        )

    else:

        st.sidebar.error(
            "Anomaly model: NOT LOADED"
        )

    if api_status.get("forecasting_model"):

        st.sidebar.success(
            "Forecasting model: LOADED"
        )

    else:

        st.sidebar.error(
            "Forecasting model: NOT LOADED"
        )

else:

    st.sidebar.error(
        "FastAPI: OFFLINE"
    )


# ============================================================
# DATA SUMMARY
# ============================================================

st.header("Process Overview")

col1, col2, col3, col4 = st.columns(4)

with col1:

    st.metric(
        "Rows",
        f"{len(filtered_df):,}",
    )

with col2:

    st.metric(
        "Equipment",
        filtered_df["equipment_id"].nunique(),
    )

with col3:

    st.metric(
        "Mean dissolved O₂",
        f"{filtered_df['dissolved_oxygen'].mean():.3f}",
    )

with col4:

    st.metric(
        "Max dissolved O₂",
        f"{filtered_df['dissolved_oxygen'].max():.3f}",
    )


# ============================================================
# PROCESS VARIABLES
# ============================================================

st.header("Process Variables")

chart_columns = [
    "temperature",
    "ph",
    "dissolved_oxygen",
    "agitation_speed",
    "pressure",
    "flow_rate",
]

available_chart_columns = [
    column
    for column in chart_columns
    if column in filtered_df.columns
]

if available_chart_columns:

    chart_df = filtered_df[
        available_chart_columns
    ].copy()

    st.line_chart(
        chart_df,
        use_container_width=True,
    )


# ============================================================
# DISSOLVED OXYGEN
# ============================================================

st.header("Dissolved Oxygen")

oxygen_df = filtered_df[
    [
        "dissolved_oxygen",
        "dissolved_oxygen_lag_1",
        "dissolved_oxygen_lag_5",
        "dissolved_oxygen_lag_30",
    ]
].copy()

st.line_chart(
    oxygen_df,
    use_container_width=True,
)


# ============================================================
# LIVE API INFERENCE
# ============================================================

st.header("AI Inference")

if filtered_df.empty:

    st.warning(
        "No data available for the selected equipment."
    )

else:

    selected_index = st.number_input(
        "Select row",
        min_value=0,
        max_value=len(filtered_df) - 1,
        value=0,
        step=1,
    )

    selected_row = filtered_df.iloc[
        selected_index
    ]

    if st.button(
        "Run AI Prediction",
        type="primary",
    ):

        if not api_status:

            st.error(
                "FastAPI is offline. "
                "Start the FastAPI server first."
            )

        else:

            try:

                result = predict(
                    selected_row
                )

                st.subheader(
                    "Prediction Result"
                )

                result_col1, result_col2, result_col3 = (
                    st.columns(3)
                )

                with result_col1:

                    st.metric(
                        "Predicted dissolved O₂",
                        f"{result['predicted_dissolved_oxygen']:.4f}",
                    )

                with result_col2:

                    st.metric(
                        "Anomaly score",
                        f"{result['anomaly_score']:.4f}",
                    )

                with result_col3:

                    if result["is_anomaly"]:

                        st.error(
                            "ANOMALY DETECTED"
                        )

                    else:

                        st.success(
                            "NORMAL"
                        )

                st.json(result)

            except Exception as exc:

                st.error(
                    f"Prediction failed: {exc}"
                )


# ============================================================
# RECENT DATA
# ============================================================

st.header("Recent Feature Data")

st.dataframe(
    filtered_df.tail(20),
    use_container_width=True,
)


# ============================================================
# FOOTER
# ============================================================

st.divider()

st.caption(
    "PHARMA — AI monitoring platform | "
    "FastAPI + MLflow + Scikit-learn + Streamlit"
)
'''

# ============================================================
# WRITE DASHBOARD
# ============================================================

DASHBOARD_FILE.write_text(
    dashboard_code,
    encoding="utf-8",
)

print()
print("Created:")
print(DASHBOARD_FILE)

# ============================================================
# SYNTAX VALIDATION
# ============================================================

print()
print("-" * 60)
print("DASHBOARD SYNTAX VALIDATION")
print("-" * 60)

syntax_check = subprocess.run(
    [
        sys.executable,
        "-m",
        "py_compile",
        str(DASHBOARD_FILE),
    ],
    cwd=ROOT,
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
)

print("Return code:", syntax_check.returncode)

if syntax_check.returncode != 0:

    print()
    print("ERROR:")
    print(syntax_check.stderr)

    raise RuntimeError(
        "Streamlit dashboard syntax validation failed."
    )

print()
print("Dashboard syntax: OK")

# ============================================================
# FINAL SETUP
# ============================================================

print()
print("=" * 60)
print("PHARMA — STREAMLIT DASHBOARD: READY")
print("=" * 60)

print()
print("Dashboard:")
print(DASHBOARD_FILE)

print()
print("FastAPI:")
print(API_URL)

print()
print("To start the dashboard, run:")
print()
print(
    f'python -m streamlit run "{DASHBOARD_FILE}"'
)

print()
print("Expected URL:")
print("http://localhost:8501")

print()
print("=" * 60)
```

```python
# ============================================================
# PHARMA — FIND ANOMALY ROWS BY EQUIPMENT
# ============================================================

from pathlib import Path
import joblib
import pandas as pd

print("=" * 60)
print("PHARMA — ANOMALIES BY EQUIPMENT")
print("=" * 60)

ROOT = Path(r"C:\Users\zosia\pharma")

DATA_FILE = ROOT / "data" / "processed" / "features.csv"
MODEL_FILE = ROOT / "models" / "anomaly_model.joblib"

ANOMALY_FEATURES = [
    "temperature",
    "ph",
    "dissolved_oxygen",
    "agitation_speed",
    "pressure",
    "flow_rate",
    "dissolved_oxygen_lag_1",
    "dissolved_oxygen_lag_2",
    "dissolved_oxygen_lag_3",
    "dissolved_oxygen_lag_5",
    "dissolved_oxygen_lag_10",
    "dissolved_oxygen_lag_15",
    "dissolved_oxygen_lag_30",
]

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

df = pd.read_csv(DATA_FILE)

print()
print("Rows:", len(df))
print("Equipment:", sorted(df["equipment_id"].unique()))

# ------------------------------------------------------------
# LOAD MODEL
# ------------------------------------------------------------

model = joblib.load(MODEL_FILE)

# ------------------------------------------------------------
# RUN ANOMALY DETECTION
# ------------------------------------------------------------

X = df[ANOMALY_FEATURES]

df["anomaly_prediction"] = model.predict(X)
df["anomaly_score"] = model.decision_function(X)
df["is_anomaly"] = df["anomaly_prediction"] == -1

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print()
print("=" * 60)
print("ANOMALIES BY EQUIPMENT")
print("=" * 60)

summary = (
    df.groupby("equipment_id")["is_anomaly"]
    .agg(
        total_rows="count",
        anomalies="sum"
    )
)

summary["anomaly_rate_%"] = (
    summary["anomalies"] / summary["total_rows"] * 100
).round(2)

print(summary)

# ------------------------------------------------------------
# SHOW ANOMALY ROWS FOR EACH EQUIPMENT
# ------------------------------------------------------------

for equipment in sorted(df["equipment_id"].unique()):

    anomalies = df[
        (df["equipment_id"] == equipment)
        & (df["is_anomaly"])
    ].copy()

    print()
    print("-" * 60)
    print(f"{equipment} — ANOMALIES")
    print("-" * 60)

    if anomalies.empty:
        print("No anomalies detected.")
        continue

    print(
        anomalies[
            [
                "timestamp",
                "equipment_id",
                "dissolved_oxygen",
                "anomaly_score",
            ]
        ].head(20).to_string()
    )

    print()
    print("ROW INDICES TO TEST:")
    print(anomalies.index.tolist()[:20])

print()
print("=" * 60)
print("DONE")
print("=" * 60)
```

```python
# ============================================================
# PHARMA — EQ-002 ROW 1924 ANOMALY DIAGNOSTIC
# ============================================================

from pathlib import Path
import joblib
import pandas as pd

ROOT = Path(r"C:\Users\zosia\pharma")

DATA_FILE = ROOT / "data" / "processed" / "features.csv"
MODEL_FILE = ROOT / "models" / "anomaly_model.joblib"

ANOMALY_FEATURES = [
    "temperature",
    "ph",
    "dissolved_oxygen",
    "agitation_speed",
    "pressure",
    "flow_rate",
    "dissolved_oxygen_lag_1",
    "dissolved_oxygen_lag_2",
    "dissolved_oxygen_lag_3",
    "dissolved_oxygen_lag_5",
    "dissolved_oxygen_lag_10",
    "dissolved_oxygen_lag_15",
    "dissolved_oxygen_lag_30",
]

print("=" * 60)
print("PHARMA — EQ-002 ROW 1924 ANOMALY DIAGNOSTIC")
print("=" * 60)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

df = pd.read_csv(DATA_FILE)

print()
print("Data rows:", len(df))
print("Data columns:", len(df.columns))

# ------------------------------------------------------------
# CHECK ROW 1924
# ------------------------------------------------------------

row_index = 1924

if row_index >= len(df):
    raise IndexError(
        f"Row {row_index} does not exist. "
        f"Dataset contains {len(df)} rows."
    )

row = df.iloc[row_index]

print()
print("-" * 60)
print("ROW INFORMATION")
print("-" * 60)

print("Index:", row_index)

if "equipment_id" in df.columns:
    print("Equipment:", row["equipment_id"])

if "timestamp" in df.columns:
    print("Timestamp:", row["timestamp"])

# ------------------------------------------------------------
# VERIFY EQUIPMENT
# ------------------------------------------------------------

if "equipment_id" in df.columns:

    if str(row["equipment_id"]) != "EQ-002":

        print()
        print("WARNING:")
        print(
            f"Row {row_index} is actually "
            f"{row['equipment_id']}, not EQ-002."
        )

# ------------------------------------------------------------
# LOAD MODEL
# ------------------------------------------------------------

print()
print("-" * 60)
print("LOADING ANOMALY MODEL")
print("-" * 60)

model = joblib.load(MODEL_FILE)

print("Model:", MODEL_FILE)
print("Model type:", type(model).__name__)

# ------------------------------------------------------------
# FEATURE CHECK
# ------------------------------------------------------------

missing = [
    feature
    for feature in ANOMALY_FEATURES
    if feature not in df.columns
]

if missing:
    raise ValueError(
        "Missing features: " + ", ".join(missing)
    )

print()
print("Feature contract: OK")

# ------------------------------------------------------------
# DISPLAY INPUT FEATURES
# ------------------------------------------------------------

X_row = row[ANOMALY_FEATURES].to_frame().T

print()
print("-" * 60)
print("INPUT FEATURES — ROW 1924")
print("-" * 60)

print(X_row.to_string(index=False))

# ------------------------------------------------------------
# CHECK MISSING VALUES
# ------------------------------------------------------------

print()
print("-" * 60)
print("MISSING VALUES")
print("-" * 60)

print(X_row.isna().sum())

# ------------------------------------------------------------
# MODEL PREDICTION
# ------------------------------------------------------------

prediction = model.predict(X_row)[0]
score = model.decision_function(X_row)[0]

is_anomaly = prediction == -1

print()
print("=" * 60)
print("MODEL RESULT")
print("=" * 60)

print("Row:", row_index)
print("Equipment:", row["equipment_id"])
print("Prediction:", prediction)
print("Anomaly score:", score)
print("Is anomaly:", is_anomaly)

print()
print("=" * 60)

if is_anomaly:
    print("RESULT: MODEL DETECTS ANOMALY")
else:
    print("RESULT: MODEL CLASSIFIES ROW AS NORMAL")

print("=" * 60)
```

```python
import pandas as pd
import joblib

from src.streaming.producer import create_stream
from src.streaming.inference import (
    ANOMALY_MODEL_FILE,
    FORECASTING_MODEL_FILE,
    ANOMALY_FEATURE_COLUMNS,
    FORECASTING_FEATURE_COLUMNS,
)

# ---------------------------------------------------------
# 1. Génération des événements
# ---------------------------------------------------------

events = list(create_stream())

print("Événements générés :", len(events))

# ---------------------------------------------------------
# 2. Conversion en DataFrame
# ---------------------------------------------------------

df_events = pd.DataFrame(events)

print("Dimensions :", df_events.shape)
display(df_events.head())
```

```python
from pathlib import Path
import joblib

PROJECT_ROOT = Path(r"C:\Users\zosia\pharma")

ANOMALY_MODEL_FILE = PROJECT_ROOT / "models" / "anomaly_model.joblib"
FORECASTING_MODEL_FILE = PROJECT_ROOT / "models" / "forecasting_model.joblib"

anomaly_model = joblib.load(ANOMALY_MODEL_FILE)
forecasting_model = joblib.load(FORECASTING_MODEL_FILE)

print("Anomaly model :", type(anomaly_model).__name__)
print("Forecasting model :", type(forecasting_model).__name__)
```

```python
# ============================================================
# BATCH INFERENCE — 4910 EVENTS
# ============================================================

# Colonnes utilisées par les modèles
ANOMALY_FEATURE_COLUMNS = [
    "temperature",
    "ph",
    "dissolved_oxygen",
    "agitation_speed",
    "pressure",
    "flow_rate",
    "dissolved_oxygen_lag_1",
    "dissolved_oxygen_lag_2",
    "dissolved_oxygen_lag_3",
    "dissolved_oxygen_lag_5",
    "dissolved_oxygen_lag_10",
    "dissolved_oxygen_lag_15",
    "dissolved_oxygen_lag_30",
]

FORECASTING_FEATURE_COLUMNS = [
    "dissolved_oxygen_lag_1",
    "dissolved_oxygen_lag_2",
    "dissolved_oxygen_lag_3",
    "dissolved_oxygen_lag_5",
    "dissolved_oxygen_lag_10",
    "dissolved_oxygen_lag_15",
    "dissolved_oxygen_lag_30",
]

# ------------------------------------------------------------
# ANOMALY DETECTION
# ------------------------------------------------------------

X_anomaly = df_events[ANOMALY_FEATURE_COLUMNS].astype(float)

df_events["anomaly_prediction"] = anomaly_model.predict(
    X_anomaly
)

df_events["anomaly_score"] = anomaly_model.decision_function(
    X_anomaly
)

df_events["is_anomaly"] = (
    df_events["anomaly_prediction"] == -1
)

# ------------------------------------------------------------
# FORECASTING
# ------------------------------------------------------------

X_forecasting = df_events[
    FORECASTING_FEATURE_COLUMNS
].astype(float)

df_events["predicted_dissolved_oxygen"] = (
    forecasting_model.predict(X_forecasting)
)

print("BATCH INFERENCE: OK")
print("Événements traités :", len(df_events))
```

```python
print("Nombre d'anomalies :", int(df_events["is_anomaly"].sum()))

print(
    "Nombre d'événements normaux :",
    int((~df_events["is_anomaly"]).sum())
)

print(
    "Taux d'anomalie :",
    round(df_events["is_anomaly"].mean() * 100, 2),
    "%"
)

print(
    "Score d'anomalie moyen :",
    round(df_events["anomaly_score"].mean(), 4)
)

print(
    "Oxygène dissous prédit — moyenne :",
    round(
        df_events["predicted_dissolved_oxygen"].mean(),
        4
    )
)

display(
    df_events[
        [
            "equipment_id",
            "timestamp",
            "anomaly_prediction",
            "is_anomaly",
            "anomaly_score",
            "predicted_dissolved_oxygen",
        ]
    ].head(10)
)
```
